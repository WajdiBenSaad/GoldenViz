# GoldenViz demo notebook

This notebook illustrates the first five Golden Rules currently implemented in GoldenViz. Each section shows a problematic chart followed by a corrected version.

```python
import matplotlib.pyplot as plt
import GoldenViz as gv

# Enable automatic reports under charts
gv.auto()
```

## Rule 1 · Clear title
A chart should have a descriptive title.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].plot([2021, 2022, 2023, 2024], [14, 16, 15, 19], marker='o')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Revenue (M€)')
axes[0].set_title('')

axes[1].plot([2021, 2022, 2023, 2024], [14, 16, 15, 19], marker='o')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Revenue (M€)')
axes[1].set_title('Revenue trend by year')

plt.tight_layout()
plt.show()
```

## Rule 2 · Axis labels
Both axes should be labeled clearly.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].bar(['Bags', 'Shoes', 'Silk'], [24, 18, 11])
axes[0].set_title('Units sold by product family')

axes[1].bar(['Bags', 'Shoes', 'Silk'], [24, 18, 11])
axes[1].set_title('Units sold by product family')
axes[1].set_xlabel('Product family')
axes[1].set_ylabel('Units sold')

plt.tight_layout()
plt.show()
```

## Rule 3 · Appropriate scale
Bar charts should usually start at zero.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
values = [95, 97, 100]
labels = ['Brand A', 'Brand B', 'Brand C']

axes[0].bar(labels, values)
axes[0].set_title('Equipment uptime comparison')
axes[0].set_xlabel('Brand')
axes[0].set_ylabel('Uptime (%)')
axes[0].set_ylim(94, 100.5)

axes[1].bar(labels, values)
axes[1].set_title('Equipment uptime comparison')
axes[1].set_xlabel('Brand')
axes[1].set_ylabel('Uptime (%)')
axes[1].set_ylim(0, 100.5)

plt.tight_layout()
plt.show()
```

## Rule 4 · Chart type
A line chart is not always the right choice for discrete categories.

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

categories = ['North', 'South', 'East', 'West']
values = [18, 11, 15, 13]

axes[0].plot(categories, values, marker='o')
axes[0].set_title('Sales by region')
axes[0].set_xlabel('Region')
axes[0].set_ylabel('Sales (k)')

axes[1].bar(categories, values)
axes[1].set_title('Sales by region')
axes[1].set_xlabel('Region')
axes[1].set_ylabel('Sales (k)')

plt.tight_layout()
plt.show()
```

## Rule 5 · Readable labels and ticks
Dense or overlapping labels hurt readability.

```python
long_labels = [
    'Very long category label 1',
    'Very long category label 2',
    'Very long category label 3',
    'Very long category label 4',
    'Very long category label 5',
    'Very long category label 6',
]
values = [3, 8, 6, 9, 4, 7]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].bar(long_labels, values)
axes[0].set_title('Customer requests by category')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Requests')

axes[1].bar(long_labels, values)
axes[1].set_title('Customer requests by category')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Requests')
axes[1].tick_params(axis='x', rotation=35)

plt.tight_layout()
plt.show()
```

## Manual mode
Automatic mode is convenient in notebooks, but you can also run GoldenViz manually.

```python
# Disable automatic reports under charts
gv.disable()

fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter([1, 2, 3, 4], [10, 14, 13, 17])
ax.set_title('Quarterly performance')
ax.set_xlabel('Quarter')
ax.set_ylabel('Performance score')
plt.show()

# Manual check
gv.check(fig)
```
